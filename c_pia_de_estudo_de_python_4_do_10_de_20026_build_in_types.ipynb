```python
entrada = input('escreva um número')
try:
   x = float(entrada) # converte a entrada em valor numerico
   if x > 0:
    print('positivo')
   elif x < 0:
    print('negativo')
   elif x == 0:
    print('zero')
except ValueError: # caso o valor digitado não seja numero ele retorna que não é numero
    print('não é numero')
```
